# RUN

```python
trial = 25

runtime_results = []
```

## Classical Test Function

```python
pop_size = 30
epoch = 100

Lshade = L_SHADE(epoch, pop_size)
CmaES = CMA_ES(epoch, pop_size)
imode = IMODE(epoch, pop_size)

winner_models = [Lshade, CmaES, imode]
```

```python
if __name__ == "__main__":

    modified_Fproblems = []
    for problem_dict in Fproblems:
        new_problem_dict = problem_dict.copy()
        if "fit_func" in new_problem_dict:
            new_problem_dict["obj_func"] = new_problem_dict.pop("fit_func")
        modified_Fproblems.append(new_problem_dict)

    start_time = time.time()
    multitask = Multitask(algorithms=winner_models, problems=modified_Fproblems, n_workers=4)
    multitask.execute(n_trials=trial, save_path=path + "history/classic", save_as="csv", save_convergence=True, verbose=True)
    runtime = time.time() - start_time
    runtime_results.append({"algorithm": "NHO", "suite": "classic", "dimension": 2, "runtime": runtime})
    print("Runtime", runtime)
```
