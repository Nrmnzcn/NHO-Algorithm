# Import Library

```python
# fundamental libraries
import os
import math
import numpy as np
import pandas as pd

# visualization
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.image as mpimg
```

# Visualization

```python
def compare_models_plot_q1(
    models=None,
    test_names=None,
    test_function=None,
    test_folder=None,
    colors=None,
    markers=None,
    out_dir=path + "figure/comparation",
    dpi=300,
    fig_width_in=3.35,     # ~85 mm (tek sütun)
    fig_height_in=2.6,     # 3 sütun sayfada dengeli yükseklik
    lw=1.3,
    ms=3.0,
    show_legend=True,
):

    os.makedirs(out_dir, exist_ok=True)

    # Q1: küçük figürde okunur fontlar
    label_fs = 9
    title_fs = 10
    tick_fs  = 9
    legend_fs = 7

    for name, function in zip(test_names, test_function):

        fig, ax = plt.subplots(figsize=(fig_width_in, fig_height_in), dpi=dpi)

        for idx, model in enumerate(models):
            csv_path = path + f"history/{test_folder}/convergence/{model}/{function}_convergence.csv"
            df = pd.read_csv(csv_path)

            label = (model.replace("Original", "")
                          .replace("Base", "")
                          .replace("OURMODEL", "Proposed")
                          .replace("ACOR", "ACO"))

            y = df.mean(axis=1)

            ax.plot(
                y,
                label=label,
                color=colors[idx] if colors is not None else None,
                marker=markers[idx] if markers is not None else None,
                linewidth=lw,
                markersize=ms,
                markevery=max(len(y)//12, 1)  # marker kalabalığını azaltır
            )

        ax.set_xlabel("Iterations", fontsize=label_fs)
        ax.set_ylabel("Fitness", fontsize=label_fs)
        ax.set_title(f"{name}", fontsize=title_fs)

        ax.tick_params(axis="both", labelsize=tick_fs)

        # Grid (Q1 için hafif)
        ax.grid(True, linewidth=0.4, alpha=0.3)

        # Legend (çok model varsa sıkışmayı önler)
        if show_legend:
            ax.legend(
                fontsize=legend_fs,
                loc="upper right",
                frameon=False,
                ncol=3,
                handlelength=1.6,
                borderaxespad=0.2
            )

        fig.tight_layout(pad=0.3)

        out_path = os.path.join(out_dir, f"{test_folder}/{name}.png")
        fig.savefig(out_path, dpi=dpi, bbox_inches="tight")
        plt.show()
        plt.close(fig)
```

## Grid - Selected Function

```python
def plot_convergence_grid_from_csv(
    models,
    names,
    functions,
    test_folder,
    colors,
    markers,
    methods,
    out_path,
    ncols=2,
    dpi=600,
    fig_width=7.2,
    fig_height_per_row=2.4,
    lw=1.2,
    ms=2.5,
    legend_ncol=5
):

    n = len(names)
    nrows = math.ceil(n / ncols)

    fig_height = fig_height_per_row * nrows + 0.8

    fig, axs = plt.subplots(
        nrows,
        ncols,
        figsize=(fig_width, fig_height),
        dpi=dpi
    )

    axs = np.array(axs).reshape(nrows, ncols)

    legend_handles = []

    for k in range(nrows * ncols):

        i, j = divmod(k, ncols)
        ax = axs[i, j]

        if k >= n:
            ax.axis("off")
            continue

        plot_title = names[k]
        function_name = functions[k]

        for idx, model in enumerate(models):

            csv_path = path + f"history/{test_folder}/convergence/{model}/{function_name}_convergence.csv"

            df = pd.read_csv(csv_path)

            label = methods[idx]

            y = df.mean(axis=1)

            line, = ax.plot(
                y,
                label=label,
                color=colors[idx],
                marker=markers[idx],
                linewidth=lw,
                markersize=ms,
                markevery=max(len(y) // 12, 1)
            )

            if k == 0:
                legend_handles.append(line)

        ax.set_title(plot_title, fontsize=10)
        ax.set_xlabel("Iterations", fontsize=9)
        ax.set_ylabel("Fitness", fontsize=9)

        ax.tick_params(axis="both", labelsize=8)
        ax.grid(True, linewidth=0.35, alpha=0.3)

    fig.legend(
        handles=legend_handles,
        labels=methods,
        loc="lower center",
        bbox_to_anchor=(0.5, 0.01),
        ncol=legend_ncol,
        fontsize=7,
        frameon=False
    )

    fig.tight_layout(rect=[0, 0.08, 1, 1])

    os.makedirs(os.path.dirname(out_path), exist_ok=True)

    fig.savefig(out_path, bbox_inches="tight", dpi=dpi)
    fig.savefig(out_path.replace(".png", ".pdf"), bbox_inches="tight")
    fig.savefig(out_path.replace(".png", ".svg"), bbox_inches="tight")

    plt.show()
    plt.close(fig)
```

#### Classical Test Function

```python
selected_functions = ["F1", "F3", "F7", "F10", "F12", "F16"]

selected_idx = [names.index(f) for f in selected_functions]

selected_names = [names[i] for i in selected_idx]
selected_csvs = [functions[i] for i in selected_idx]


plot_convergence_grid_from_csv(
    models=MODELS,
    names=selected_names,
    functions=selected_csvs,
    test_folder="classic",
    colors=colors,
    markers=markers,
    methods=methods,
    out_path=path + "figure/grids/classic_selected_grid.pdf",
    ncols=2
)
```
